### Eigenschaften von Algorithmen

> Ein Algorithmus muss **vier Eigenschaften** erfüllen. Diese Fragen helfen beim Prüfen:
>
> - **Endlichkeit:** Kommt der Ablauf nach endlich vielen Schritten zu einem Ende?
> - **Ausführbarkeit:** Kann der Ausführende jeden einzelnen Schritt verstehen und umsetzen?
> - **Eindeutigkeit:** Ist immer klar, welcher Schritt als Nächstes kommt? Liefern gleiche Eingaben immer dasselbe Ergebnis?
> - **Allgemeingültigkeit:** Funktioniert es für alle Aufgaben dieser Art, nicht nur für ein einzelnes Beispiel?

**Programm A: Geheimschrift**

````python
# Programm A: Geheimschrift
alphabet = "abcdefghijklmnopqrstuvwxyz"
text = input("Deine Nachricht: ")
schluessel = input("Schlüssel (Zahl von 1 bis 25): ")
while not (schluessel.isdigit() and 1 <= int(schluessel) <= 25):
    schluessel = input("Bitte eine Zahl von 1 bis 25: ")
verschiebung = int(schluessel)

geheim = ""
for zeichen in text:
    klein = zeichen.lower()
    if klein in alphabet:
        neu = alphabet[(alphabet.index(klein) + verschiebung) % 26]
        if zeichen.isupper():
            neu = neu.upper()
        geheim = geheim + neu
    else:
        geheim = geheim + zeichen

print("Verschlüsselt: " + geheim)
print("Zum Entschlüsseln: Geheimtext mit dem Schlüssel " + str(26 - verschiebung) + " noch einmal eingeben.")
````

**Programm C: Wochentag-Rechner**

````python
# Programm C: Wochentag-Rechner
def zahl_eingeben(frage, kleinste, groesste):
    eingabe = input(frage)
    while not (eingabe.isdigit() and kleinste <= int(eingabe) <= groesste):
        eingabe = input("Bitte eine Zahl von " + str(kleinste) + " bis " + str(groesste) + ": ")
    return int(eingabe)

print("An welchem Wochentag wurdest du geboren?")
jahr = zahl_eingeben("Jahr: ", 1583, 9999)
monat = zahl_eingeben("Monat (1–12): ", 1, 12)
schaltjahr = (jahr % 4 == 0 and jahr % 100 != 0) or jahr % 400 == 0
tage_im_monat = [31, 29 if schaltjahr else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
tag = zahl_eingeben("Tag: ", 1, tage_im_monat[monat - 1])

# Wochentagsformel von Christian Zeller (1882)
m, j = monat, jahr
if m < 3:
    m = m + 12
    j = j - 1
h = (tag + (13 * (m + 1)) // 5 + j % 100 + (j % 100) // 4 + (j // 100) // 4 + 5 * (j // 100)) % 7
wochentage = ["Samstag", "Sonntag", "Montag", "Dienstag", "Mittwoch", "Donnerstag", "Freitag"]

print("Der " + str(tag) + "." + str(monat) + "." + str(jahr) + " war ein " + wochentage[h] + ".")
````

**Programm D: Primfaktorzerlegung**

````python
# Programm D: Primfaktorzerlegung
eingabe = input("Gib eine ganze Zahl größer als 1 ein (zum Beispiel 360 oder 123456789): ")
while not (eingabe.isdigit() and 1 < int(eingabe) <= 10**12):
    eingabe = input("Bitte eine ganze Zahl von 2 bis 1 000 000 000 000: ")
zahl = int(eingabe)

rest = zahl
teiler = 2
faktoren = []
while teiler * teiler <= rest:
    while rest % teiler == 0:
        faktoren.append(teiler)
        rest = rest // teiler
    teiler = teiler + 1
if rest > 1:
    faktoren.append(rest)

if len(faktoren) == 1:
    print(str(zahl) + " ist eine Primzahl!")
else:
    print(str(zahl) + " = " + " · ".join(str(f) for f in faktoren))
````

**Programm E: Passwort-Generator** (mehrmals mit derselben Eingabe ausführen)

### Der Türsteher: Eingaben prüfen

Ein Escape-Room ist für 2 bis 6 Personen buchbar. Das Buchungsprogramm rechnet den Preis aus. Führe die Zelle aus und gib 9 ein.

````python
personen = int(input("Wie viele Personen (2 bis 6)? "))
print("Gebucht für " + str(personen) + " Personen. Preis: " + str(personen * 25) + " Euro")
````

Das Programm bucht auch 9 Personen, obwohl der Raum nur für 6 reicht. Eine `while`-Schleife kann wie ein Türsteher arbeiten: Wer die Bedingung nicht erfüllt, muss zurück und es noch einmal versuchen. Führe die Zelle aus und gib zuerst ungültige Zahlen ein.

````python
personen = int(input("Wie viele Personen (2 bis 6)? "))
while personen < 2 or personen > 6:
    print("Der Raum ist für 2 bis 6 Personen buchbar. Bitte noch einmal.")
    personen = int(input("Wie viele Personen (2 bis 6)? "))
print("Gebucht für " + str(personen) + " Personen. Preis: " + str(personen * 25) + " Euro")
````

> **Eingaben prüfen** geht immer nach demselben Muster:
>
> 1. Wert einlesen.
> 2. **Solange** der Wert **ungültig** ist: Hinweis ausgeben und den Wert **erneut einlesen**.
> 3. Nach der Schleife ist der Wert sicher gültig und wird verarbeitet.
>
> Die Bedingung der Schleife beschreibt also die **ungültigen** Eingaben. Ungültig sind hier alle Zahlen kleiner als 2 **oder** größer als 6.

### Rechnen, bis es reicht

Ein Blatt Papier ist etwa 0,1 mm dick. Faltet man es einmal, ist es doppelt so dick. Wie oft müsste man es falten, damit der Stapel bis zum Mond reicht (384 400 km)? Schätze zuerst!

Diesmal hört die Schleife nicht wegen einer Eingabe auf, sondern weil ein berechneter Wert groß genug ist. Führe die Zelle aus.

````python
dicke = 0.1               # in Millimetern
anzahl = 0
mond = 384400 * 1000000   # Entfernung zum Mond in Millimetern
while dicke < mond:
    dicke = dicke * 2
    anzahl = anzahl + 1
print(anzahl)
````

Ändere das Programm: Wie oft muss man falten, bis der Stapel 400 km hoch ist, so hoch wie die Raumstation ISS fliegt?

> **Werte in der Schleife verändern:** Eine Variable bekommt **vor** der Schleife einen Startwert (`dicke = 0.1`). In jedem Durchlauf wird ihr neuer Wert aus dem alten berechnet (`dicke = dicke * 2`). Eine zweite Variable **zählt** die Durchläufe (`anzahl = anzahl + 1`). Vor jedem Durchlauf prüft Python die Bedingung mit dem neuen Wert.
>
> Ändert sich im Schleifenkörper nichts, was die Bedingung beeinflusst, hört die Schleife nie auf (**Endlosschleife**). Eine laufende Zelle stoppst du mit ■ in der Werkzeugleiste.

Auch der Zufall kann bestimmen, wann eine Schleife endet. `import random` lädt die Zufallsfunktionen von Python (mehr dazu in Kapitel 4), und `random.randint(1, 6)` liefert eine zufällige Zahl von 1 bis 6, wie ein Würfel. Führe die Zelle mehrmals aus. Gibt es eine größte Anzahl von Würfen, die man für eine Sechs braucht?

````python
import random

wurf = 0
versuche = 0
while wurf != 6:
    wurf = random.randint(1, 6)
    versuche = versuche + 1
    print("Wurf: " + str(wurf))
print("Eine Sechs nach " + str(versuche) + " Würfen!")
````

### Ungültige Parameter: der Fehlerwert

Was soll eine Funktion tun, wenn sie mit unsinnigen Werten aufgerufen wird? Führe die Zelle aus.

````python
def eintritt(alter):
    if alter < 0:
        print("Fehler: Das Alter kann nicht negativ sein!")
        return -1
    if alter < 6:
        return 0
    elif alter < 18:
        return 5
    else:
        return 9

print(eintritt(4))
print(eintritt(15))
print(eintritt(40))
print(eintritt(-3))
````

> Eine Funktion kann mehrere `return` haben. Das **erste** `return`, das erreicht wird, beendet die Funktion **sofort**, alle Zeilen danach werden übersprungen.
>
> **Fehlerwert:** Bei ungültigen Parametern gibt die Funktion eine Fehlermeldung aus und gibt einen Wert zurück, der als echtes Ergebnis nie vorkommt, hier `-1`. So stürzt das Programm nicht ab, und wer die Funktion aufruft, kann am Rückgabewert erkennen, dass etwas schiefgegangen ist.

### Muster prüfen mit `re.fullmatch`

Führe die Zelle aus. Ändere dann `eingabe` nacheinander in `"fünfzig"`, `"007"`, `"4 2"` und `""`. Wann passt das Muster?

````python
import re

eingabe = "42"
if re.fullmatch("[0-9]+", eingabe):
    print("passt: nur Ziffern")
else:
    print("passt nicht")
````

> Ein **regulärer Ausdruck** ist ein Muster aus Bausteinen:
>
> - `[0-9]` steht für **ein** Zeichen aus dem Bereich 0 bis 9, also eine Ziffer. Andere Bereiche gehen auch, zum Beispiel `[1-6]`.
> - `[a-z]` steht für einen Kleinbuchstaben, `[A-Z]` für einen Großbuchstaben (jeweils ohne Umlaute und ß).
> - In eine eckige Klammer dürfen mehrere Bereiche und einzelne Zeichen: `[a-zA-Z]` ist ein Klein- **oder** Großbuchstabe, `[a-zA-ZäöüÄÖÜß]` erlaubt auch Umlaute und ß, `[0-9,]` eine Ziffer oder ein Komma. Auch dann steht die ganze Klammer immer für **ein** Zeichen.
> - `+` hinter einem Baustein: Das Zeichen kommt **ein- oder mehrmals** vor.
> - `{5}` hinter einem Baustein: Das Zeichen kommt **genau 5-mal** vor. `{2,4}` heißt 2- bis 4-mal.
> - Buchstaben, Ziffern und Zeichen wie `-`, `:` oder ein Leerzeichen außerhalb der eckigen Klammern stehen für sich selbst.
>
> `re.fullmatch(muster, text)` aus der Bibliothek `re` prüft, ob der **ganze** Text zum Muster passt. In einer Bedingung zählt ein Treffer als wahr, kein Treffer als falsch.

Umlaute gehören nicht zum Bereich `a-z`. Führe die Zelle aus und vergleiche die beiden Muster. `bool(…)` wandelt das Ergebnis in `True` oder `False` um.

````python
import re

print(bool(re.fullmatch("[a-z]+", "größe")))
print(bool(re.fullmatch("[a-zäöüß]+", "größe")))
````

#### Übung 5.2: Regex-Golf

Beim Golf gewinnt, wer mit möglichst wenigen Schlägen einlocht. Beim **Regex-Golf** suchst du ein möglichst kurzes Muster, das alle Wörter einer Gruppe trifft und keines der anderen. Führe zuerst die eingeklappte Zelle mit dem Spielfeld aus. Ändere dann das Muster in den Level-Zellen, bis du „Geschafft!“ liest.

````python
# Spielfeld für Regex-Golf. Diese Zelle einfach ausführen, den Code musst du nicht verstehen.
import re

def _pruefe(muster, treffen, nicht_treffen):
    fehler = 0
    for wort in treffen:
        if re.fullmatch(muster, wort):
            print("✓ " + wort)
        else:
            print("✗ " + wort + "  (sollte passen)")
            fehler = fehler + 1
    for wort in nicht_treffen:
        if re.fullmatch(muster, wort):
            print("✗ " + wort + "  (sollte NICHT passen)")
            fehler = fehler + 1
        else:
            print("✓ " + wort + "  passt nicht, richtig so")
    if fehler == 0:
        print("Geschafft! Dein Muster hat " + str(len(muster)) + " Zeichen.")
    else:
        print(str(fehler) + " Fehler, versuche es noch einmal.")

def level1(muster):
    _pruefe(muster, ["01069", "10557", "80331"], ["1069", "010699", "0106a"])

def level2(muster):
    _pruefe(muster, ["07:45", "18:30", "00:00"], ["7:45", "0745", "07-45"])

def level3(muster):
    _pruefe(muster, ["Anna", "Ben", "Mustafa"], ["anna", "BEN", "Ben2"])

def level4(muster):
    _pruefe(muster, ["DD-AB 123", "DD-X 7", "L-MK 2024"], ["DD-ab 123", "DDAB 123", "DD-ABC 1"])

print("Spielfeld bereit.")
````

**Level 1:** Postleitzahlen. Treffen: `01069`, `10557`, `80331`. Nicht treffen: `1069`, `010699`, `0106a`.

````python
level1("[0-9]+")
````

**Level 2:** Uhrzeiten. Treffen: `07:45`, `18:30`, `00:00`. Nicht treffen: `7:45`, `0745`, `07-45`.

````python
level2("")
````

**Level 3:** Vornamen. Treffen: `Anna`, `Ben`, `Mustafa`. Nicht treffen: `anna`, `BEN`, `Ben2`.

````python
level3("")
````

★ **Level 4:** Kfz-Kennzeichen. Treffen: `DD-AB 123`, `DD-X 7`, `L-MK 2024`. Nicht treffen: `DD-ab 123`, `DDAB 123`, `DD-ABC 1`.

### Eingaben prüfen mit `while`

Jetzt verbindest du reguläre Ausdrücke mit dem Türsteher aus Kapitel 3. Die Schleife prüft diesmal nicht, ob eine Zahl in einem Bereich liegt, sondern ob die Eingabe die richtige **Form** hat. Führe die Zelle aus und gib zuerst ungültige Postleitzahlen ein.

````python
import re

plz = input("Postleitzahl: ")
while not re.fullmatch("[0-9]{5}", plz):
    print("Ungültig! Eine Postleitzahl hat genau 5 Ziffern.")
    plz = input("Postleitzahl: ")
print("Gespeichert: " + plz)
````

> **Eingaben prüfen mit regulären Ausdrücken:** Die Schleife läuft, solange die Eingabe **nicht** zum Muster passt: `while not re.fullmatch(muster, eingabe):`. Erst nach der Schleife wird die Eingabe weiterverarbeitet, zum Beispiel mit `int` umgewandelt. So kann kein `ValueError` mehr entstehen.

### Suchen und Ersetzen

Mit Strg + F findest du in einem Text ein bestimmtes Wort. Mit einem Muster findest du dagegen **alle** Stellen, die eine bestimmte Form haben, zum Beispiel alle Uhrzeiten, ohne sie vorher zu kennen. Führe die Zelle aus. Ersetze dann das Muster nacheinander durch `"[0-9]{5}"` und `"[0-9]{2}:[0-9]{2}"`.

````python
import re

notiz = "Treffen 07:45 am Bahnhof. Notfall: 0176 5551234. Unterkunft: 10557 Berlin, Zimmer 12. Rückfahrt 18:30."
print(re.findall("[0-9]+", notiz))
````

Mit einem Muster kann man auch ersetzen, zum Beispiel Telefonnummern schwärzen, bevor man einen Text weitergibt:

````python
import re

text = "Ruf mich an: 0176 5551234. Oder Tom: 0151 7654321."
print(re.sub("[0-9]+ [0-9]+", "(Nummer gelöscht)", text))
````

> - `re.findall(muster, text)` findet **alle** Stellen im Text, die zum Muster passen, auch mitten in längeren Wörtern oder Zahlen. Die Treffer stehen in eckigen Klammern, durch Kommas getrennt. Zum **Prüfen** einer Eingabe nimmt man deshalb `re.fullmatch`.
> - `re.sub(muster, ersatz, text)` ersetzt alle passenden Stellen durch `ersatz` (sub von englisch substitute, ersetzen).
>
> Auch in VS Code kannst du mit regulären Ausdrücken suchen und ersetzen: Strg + H öffnet das Ersetzen-Feld, das Symbol `.*` schaltet reguläre Ausdrücke ein.
